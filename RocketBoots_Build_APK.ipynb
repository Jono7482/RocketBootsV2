# 🚀 Rocket Boots - 1-Click Android APK Builder
This Google Colab notebook automatically builds a modern **64-bit Android APK (Android 14+ / API 34)** for your **Rocket Boots** game.

### How to use:
1. Open this notebook in [Google Colab](https://colab.research.google.com/).
2. Click **Runtime** > **Run all** (or press `Ctrl+F9`).
3. In Step 2, upload the `RocketBoots-v0.5.0-Source.zip` file.
4. Colab will download the Android SDK/NDK and build the APK in ~10 minutes.
5. Once done, the newly compiled APK will automatically download to your phone or PC!

### Step 2: Upload and Extract RocketBoots-v0.5.0-Source.zip

```python
import os, zipfile, glob
from google.colab import files

if not os.path.exists("main.py"):
    print("Please upload RocketBoots-v0.5.0-Source.zip:")
    uploaded = files.upload()
    for name in uploaded.keys():
        if name.endswith(".zip"):
            with zipfile.ZipFile(name, "r") as z:
                z.extractall(".")
    print("Project extracted successfully!")
else:
    print("main.py already present in workspace!")
```

### Step 3: Compile Android APK (API 34, 64-bit ARM)

```python
!buildozer -v android debug
```

### Step 4: Download the Compiled APK

```python
apks = glob.glob("bin/*.apk")
if apks:
    latest_apk = max(apks, key=os.path.getctime)
    print(f"Build complete! Triggering download for {latest_apk}...")
    files.download(latest_apk)
else:
    print("Error: No APK found in bin/ folder. Please check build logs above.")
```
