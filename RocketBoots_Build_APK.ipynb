# 🚀 Rocket Boots - 1-Click Android APK Builder
This Google Colab notebook builds a modern **64-bit Android APK (Android 14+ / API 34)** for the **Rocket Boots** game.

### Instructions:
1. In the top menu, click **Runtime** > **Run all** (or press `Ctrl+F9`).
2. If prompted in Step 2, upload the `RocketBoots-v0.5.0-Source.zip` file.
3. Wait ~8–12 minutes while Google Colab compiles the Android APK.
4. The compiled `.apk` will automatically download to your computer or phone!

### Step 1: Install System Dependencies & Build Tools

### Step 2: Upload and Extract Project Files

```python
import os, zipfile, glob
from google.colab import files

if not os.path.exists("main.py") or not os.path.exists("buildozer.spec"):
    print("Please upload RocketBoots-v0.5.0-Source.zip:")
    uploaded = files.upload()
    for name in uploaded.keys():
        if name.endswith(".zip"):
            with zipfile.ZipFile(name, "r") as z:
                z.extractall(".")
    print("Project files extracted successfully!")
else:
    print("Project files already present!")
```

### Step 3: Build 64-bit Android APK

```python
import os
os.environ['ACLOCAL_PATH'] = '/usr/share/aclocal'
!export ACLOCAL_PATH=/usr/share/aclocal && yes | buildozer -v android debug
```

### Step 4: Download Your Modern APK

```python
apks = glob.glob("bin/*.apk")
if apks:
    latest_apk = max(apks, key=os.path.getctime)
    print(f"Build complete! Downloading {latest_apk}...")
    files.download(latest_apk)
else:
    print("Error: No APK found in bin/ folder. Please check the build output in Step 3 above.")
```
